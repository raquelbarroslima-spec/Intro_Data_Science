5.2 - O valor total das despesas foi de R$\$$ 3,401.40 e a média de gastos foi de R$\$$ 1,700.70

5.3 - O valor da despesa com Energia Elétrica foi R$\$$ 1.200,80 e a despesa com Saúde foi R$\$$ 2.200,60

5.4 - O valor total das despesas foi de R$\$$ 3.401,40 e a média de gastos foi de R$\$$ 1.700,70

```python
energia = 1200.80
saude = 2200.60

# 5.1 - Formato americano
print(f"O valor da despesa com Energia Elétrica foi R$ {energia:,.2f} e a despesa com Saúde foi R$ {saude:,.2f}")

# 5.2 - Total e média formato americano
total = energia + saude
media_gastos = total / 2
print(f"O valor total das despesas foi de R$ {total:,.2f} e a média de gastos foi de R$ {media_gastos:,.2f}")

# 5.3 - Formato brasileiro (vírgula como decimal e ponto como milhar)
def formato_br(valor):
    return f"{valor:,.2f}".replace(',', 'X').replace('.', ',').replace('X', '.')

print(f"O valor da despesa com Energia Elétrica foi R$ {formato_br(energia)} e a despesa com Saúde foi R$ {formato_br(saude)}")

# 5.4 - Total e média formato brasileiro
print(f"O valor total das despesas foi de R$ {formato_br(total)} e a média de gastos foi de R$ {formato_br(media_gastos)}")
```

#### 6 - Crie linhas de código em que o usuário deve informar apenas os números do CPF (que contém 11 caracteres). Nessas linhas de código você precisa criar um alerta de erro caso o usuário preencha os caracteres "." ou "-". Depois disso crie um código que mostre na tela o CPF no formato tradicional (xxx.xxx.xxx-xx). Resumindo: O usuário precisa informar apenas os números do CPF(xxxxxxxxxxx), mas, no momento da exibição na tela o CPF informado deve estar no formato xxx.xxx.xxx-xx.

```python
print("=== CADASTRO DE CPF ===")
print("Digite apenas os 11 números do CPF (sem pontos ou traços)\n")

cpf = input("Informe o CPF (apenas números): ").strip()

# Verificação de caracteres inválidos
if '.' in cpf or '-' in cpf:
    print("\n✗ ERRO: O CPF não deve conter pontos (.) ou traços (-)!")
    print("Digite apenas os 11 números.")
else:
    # Verificação de quantidade de caracteres
    if len(cpf) == 11 and cpf.isdigit():
        cpf_formatado = f"{cpf[:3]}.{cpf[3:6]}.{cpf[6:9]}-{cpf[9:]}"
        print(f"\nCPF cadastrado com sucesso: {cpf_formatado}")
    else:
        print("\n✗ ERRO: O CPF deve conter exatamente 11 dígitos numéricos!")
```

7 - Crie linhas de código em que o usuário deve informar apenas os números do CPF (que contém 11 caracteres). Neste caso o usuário poderá informar o CPF no formato que quiser, contanto que ao final haja uma codificação de CPF válida (11 caracteres numéricos).

Resumindo: O usuário pode informar o CPF no formato que desejar, mas, no momento da exibição na tela o CPF informado deve estar no formato xxx.xxx.xxx-xx. Caso o que o usuário informar não resulte no formato padrão dos CPF (xxx.xxx.xxx-xx) uma mensagem de erro deverá ser informada na tela.

```python
print("=== CADASTRO DE CPF ===")
print("Você pode digitar o CPF no formato que desejar\n")

cpf_input = input("Informe o CPF: ").strip()

# Remove todos os caracteres não numéricos
cpf_numeros = ''.join(filter(str.isdigit, cpf_input))

# Verificação do CPF
if len(cpf_numeros) == 11:
    cpf_formatado = f"{cpf_numeros[:3]}.{cpf_numeros[3:6]}.{cpf_numeros[6:9]}-{cpf_numeros[9:]}"
    print(f"\n✓ CPF válido: {cpf_formatado}")
else:
    print(f"\n✗ ERRO: CPF inválido!")
    print(f"Foram encontrados {len(cpf_numeros)} dígitos numéricos.")
    print("Um CPF válido deve conter exatamente 11 dígitos numéricos.")
```

```python
print("=== CADASTRO DE NÚMERO DE CELULAR ===")
print("Formato padrão: 9xxxx-xxxx")
print("Você pode informar com ou sem o primeiro dígito '9'\n")

celular_input = input("Informe o número do celular: ").strip()

# Remove caracteres não numéricos
celular_numeros = ''.join(filter(str.isdigit, celular_input))

# Verificações
if len(celular_numeros) == 9:
    # Usuário informou sem o 9 inicial, vamos adicionar
    celular_numeros = '9' + celular_numeros
    celular_formatado = f"{celular_numeros[:5]}-{celular_numeros[5:]}"
    print(f"\n✓ Celular cadastrado com sucesso: {celular_formatado}")

elif len(celular_numeros) == 10:
    # Verifica se começa com 9
    if celular_numeros[0] == '9':
        celular_formatado = f"{celular_numeros[:5]}-{celular_numeros[5:]}"
        print(f"\n✓ Celular cadastrado com sucesso: {celular_formatado}")
    else:
        print("\n✗ ERRO: O número de celular deve começar com o dígito 9!")

else:
    print("\n✗ ERRO: Número de celular inválido!")
    if len(celular_numeros) < 9:
        print("O número informado possui menos de 9 dígitos.")
    elif len(celular_numeros) > 10:
        print("O número informado possui mais de 10 dígitos.")
    else:
        print("Formato incorreto. O celular deve ter 9 ou 10 dígitos numéricos.")
```

2 - Crie linhas de código que permitam que o usuário informe o valor do seu salário e de suas despesas no padrão monetário brasileiro (por exemplo Salário = 5.520,45 e Despesas = 3.895,86). A partir dessas informações seu sistema deve apurar o "Resultado Financeiro" (que é o resultado da diferença entre o valor do "Salário" e o valor das "Despesas") e também apurar a "Situação Financeira" (que será considerada "Boa", quando o Resultado Financeiro for maior ou igual a zero, e "Ruim", quando o Resultado Financeiro for negativo).

Ao final o seu sistema deve mostrar na tela as seguintes informações:

Salário: R$\$$ 5.520,45

Despesas: R$ 3.895,86

Resultado Financeiro: R$\$$ 1.624,59

Situação Financeira: Boa

```python
print("=== SISTEMA FINANCEIRO ===\n")

# Função para converter formato brasileiro para float
def converter_para_float(valor_str):
    # Remove possíveis espaços
    valor_str = valor_str.strip()
    # Substitui vírgula por ponto (formato americano)
    valor_str = valor_str.replace('.', '').replace(',', '.')
    return float(valor_str)

# Função para formatar em padrão brasileiro
def formato_br(valor):
    return f"{valor:,.2f}".replace(',', 'X').replace('.', ',').replace('X', '.')

# Entrada de dados
salario_input = input("Informe o valor do Salário (ex: 5.520,45): R$ ")
despesas_input = input("Informe o valor das Despesas (ex: 3.895,86): R$ ")

# Conversão
salario = converter_para_float(salario_input)
despesas = converter_para_float(despesas_input)

# Cálculos
resultado_financeiro = salario - despesas

# Determinação da situação
if resultado_financeiro >= 0:
    situacao = "Boa"
else:
    situacao = "Ruim"

# Exibição dos resultados
print(f"\n{'='*40}")
print(f"Salário: R$ {formato_br(salario)}")
print(f"Despesas: R$ {formato_br(despesas)}")
print(f"Resultado Financeiro: R$ {formato_br(resultado_financeiro)}")
print(f"Situação Financeira: {situacao}")
print(f"{'='*40}")
```
